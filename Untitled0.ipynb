```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import PassiveAggressiveClassifier

data = {
    'text': [
        "The government announced a new policy for free electricity to all citizens starting next month.",
        "Aliens landed in New York City and shook hands with the mayor yesterday evening.",
        "Scientists discover a new planet that could potentially support human life in our solar system.",
        "Drinking boiled water with lemon cures all types of cancer instantly within two days.",
        "Local school students win national science competition with their new eco-friendly robot model."
    ],
    'label': ['REAL', 'FAKE', 'REAL', 'FAKE', 'REAL']
}

df = pd.DataFrame(data)

X = df['text']
y = df['label']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.4, random_state=42)

vectorizer = TfidfVectorizer(stop_words='english')
X_train_vec = vectorizer.fit_transform(X_train)
X_test_vec = vectorizer.transform(X_test)

model = PassiveAggressiveClassifier(max_iter=50)
model.fit(X_train_vec, y_train)

print("✨ मॉडेल तयार झाले आहे!")
```

```python
user_input = [input(" Enter करा: ")]

user_vec = vectorizer.transform(user_input)

prediction = model.predict(user_vec)

print("----------------------------------------")
print(" (Result):", prediction[0])
print("----------------------------------------")
```
